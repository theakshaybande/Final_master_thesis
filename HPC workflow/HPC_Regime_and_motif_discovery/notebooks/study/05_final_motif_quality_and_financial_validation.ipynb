## Limitations

The analysis is unsupervised and has no labelled financial motif ground truth. Matrix Profile pairs and LoCoMotif sets are different motif objects. Small event counts, especially Matrix Profile pairs, are descriptive only when `inference_status` is `insufficient_n`. These results do not establish predictability, profitability, or causality.
